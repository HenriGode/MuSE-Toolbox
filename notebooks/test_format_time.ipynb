# Format Time Testing
This notebook demonstrates the capabilities of the newly added `format_time` function in `muse_toolbox.utils.format_utils`, including the `detailed` and `levels` parameters.

```python
import sys
from pathlib import Path

# Ensure src is in the path to import muse_toolbox
src_path = str(Path().resolve().parent / 'src')
if src_path not in sys.path:
    sys.path.append(src_path)

from muse_toolbox.utils.format_utils import format_time
```

### Standard Test Cases (Default levels=3)
Below we define multiple test times across the full scale from weeks to nanoseconds.

```python
test_cases = [
    # Weeks and Days
    3 * 604800 + 2 * 86400 + 5 * 3600 + 24 * 60,  # 3 weeks 2 days 5 hours (truncates mins)
    3 * 604800 + 2 * 86400 + 24 * 60,             # 3 weeks 2 days 24 mins
    1 * 604800 + 1 * 86400,                       # 1 week 1 day (singular labels)
    604800.0,                                     # Exactly 1 week

    # Hours, Mins, and Seconds (Decimals and Integers)
    5 * 3600 + 2 * 60 + 3,                        # 5 hours 2 mins 3 s (no decimals for 3rd level s)
    2 * 60 + 3.3,                                 # 2 mins 3.3 s (decimals allowed in 2nd level)
    60 + 1.123,                                   # 1 min 1.123 s
    3.300,                                        # 3.3 s
    
    # Sub-second measurements
    0.5,                                          # 500 ms
    0.0005,                                       # 500 us
    0.0000005,                                    # 500 ns
]

print("=== DETAILED MODE (detailed=True, levels=3) ===")
for t in test_cases:
    print(f"{t:<15.4f} -> {format_time(t, detailed=True, levels=3)}")

print("\n=== STANDARD MODE (detailed=False) ===")
for t in test_cases:
    print(f"{t:<15.4f} -> {format_time(t, detailed=False)}")
```

### Edge Cases & Custom Levels
Testing `levels` from 1 to 6 on particularly weird boundaries.

```python
edge_cases = [
    604799.999,                                   # Just under 1 week (6 days 23 hours 59 mins 59.999 s)
    604800.001,                                   # Just over 1 week (1 week 0 days 0 hours 0 mins 0.001 s)
    86400.999,                                    # 1 day 0 hours 0 mins 0.999 s
    3599.999,                                     # 59 mins 59.999 s
    1.001,                                        # 1.001 s
    0.999,                                        # 999 ms (sub-second bypasses detailed mode entirely)
    3 * 604800 + 0 * 86400 + 5 * 3600 + 0 * 60 + 0.5, # 3 weeks 0 days 5 hours 0 mins 0.5 s
]

for t in edge_cases:
    print(f"\n--- Time: {t} seconds ---")
    for lvl in range(1, 7):
        print(f"Levels = {lvl}: {format_time(t, detailed=True, levels=lvl)}")
```
