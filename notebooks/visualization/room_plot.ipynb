```python
#%%
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib as mpl

# --- LaTeX/PGF Setup for Paper-Quality Figures ---
mpl.use("pgf")
mpl.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
})

# Room dimensions
room_length = 7  # in meters
room_width = 6  # in meters

# --- Figure and Font Configuration ---
# 1 inch = 72 points. LaTeX column width is 244.6937 pt.
columnwidth_in = 244.6937 / 72
fig_width = columnwidth_in
# Adjust height for a good aspect ratio. Since the room is roughly square,
# a height close to the width might be appropriate.
fig_height = fig_width * (room_width / room_length)

# Font sizes in points (pt)
label_fontsize = 9
tick_fontsize = 9
legend_fontsize = 9
text_fontsize = 9 # For text inside the plot

# Linewidth
LW = 1.5



# Loudspeaker positions (x, y)
loudspeakers = np.linspace(-180,150,12)/180*np.pi
loudspeakers = np.array([0, 30, -90, -120])/180*np.pi

# Microphone positions (x, y)
dist = 0.15
lsdist = 2
microphones_x = [-2.5, -2.5, 2.5, 2.5]#, -1.5, 1.5]
microphones_y = [0.5, -0.5, 0.5, -0.5]#, 0, 0]

colors = sns.color_palette("colorblind").as_hex()

color_m = colors[0]
color_n = colors[3]
color_h = colors[1]
color_s = colors[2]

# config makersizes
maker_size_LS = 40
maker_size_mics = 10

# Plot the room
fig, ax = plt.subplots(figsize=(fig_width, fig_height), constrained_layout=True)
ax.set_xlim(0, room_length)
ax.set_ylim(0, room_width)

t = np.linspace(0,2*np.pi, 1000)
x = np.linspace(-1,1, 1000)

rad = 0.3 # Size of head for both dummy head and sources

# Plot loudspeakers as heads (circles with nose pointing to center)
cx = room_length / 2
cy = room_width / 2

for i, ls in enumerate(loudspeakers):
    # Source position (same logic as before)
    sx = cx + np.sin(ls) * lsdist
    sy = cy + np.cos(ls) * lsdist
    
    # Vector from source to center (direction for nose)
    dx = cx - sx
    dy = cy - sy
    norm_val = np.sqrt(dx*dx + dy*dy)
    fx = dx / norm_val 
    fy = dy / norm_val
    
    # Right vector (perpendicular to forward)
    # If forward is (0,1) [North], Right is (1,0) [East].
    rx = fy
    ry = -fx
    
    # Calculate rotated nose points
    # Original nose: x points Right, y points Forward
    nose_x_pts = sx + rx * (x * rad / 4) + fx * (-np.abs(x) * rad / 2 + 1.5 * rad)
    nose_y_pts = sy + ry * (x * rad / 4) + fy * (-np.abs(x) * rad / 2 + 1.5 * rad)
    
    # Circle points
    circ_x = sx + rad * np.cos(t)
    circ_y = sy + rad * np.sin(t)
    
    kw = {'c': color_s, 'linewidth': LW}
    
    if i == 0:
        ax.plot(circ_x, circ_y, **kw)
    else:
        ax.plot(circ_x, circ_y, **kw)
    ax.plot(nose_x_pts, nose_y_pts, **kw)
    
    # Add labels for sources just behind the speaker's head
    label_x = sx - fx * 0.55
    label_y = sy - fy * 0.55
    ax.text(label_x, label_y, rf'Source {i+1}', ha='center', va='center', fontsize=text_fontsize, color=color_s)
    
    # Add symbolic soundwaves advancing to the center (curves facing the listener)
    theta_wave = np.linspace(-np.pi/5, np.pi/5, 50)
    for wave_dist in [0.55, 0.8, 1.05]:
        wave_x = sx + wave_dist * (fx * np.cos(theta_wave) + rx * np.sin(theta_wave))
        wave_y = sy + wave_dist * (fy * np.cos(theta_wave) + ry * np.sin(theta_wave))
        ax.plot(wave_x, wave_y, color=color_s, linewidth=LW*0.6, alpha=0.8)

# Plot dummy head
ax.plot(rad*np.cos(t)+room_length/2, rad*np.sin(t)+room_width/2, c=color_h, linewidth=LW)
ax.plot(x*rad/4+room_length/2, -abs(x)*rad/2+room_width/2+1.5*rad, c=color_h, linewidth=LW)

plt.text(3, 2.35, r'Listener', fontsize=text_fontsize, c=color_h)

# Plot microphones as unfilled red circles
ax.scatter(room_length/2+dist*microphones_x[0], room_width/2+dist*microphones_y[0], marker='o', edgecolors=color_m, facecolors='none', s=maker_size_mics, linewidths=LW, label=r'Microphones')
for mic_x, mic_y in zip(microphones_x[1:], microphones_y[1:]):
    ax.scatter(room_length/2+dist*mic_x, room_width/2+dist*mic_y, marker='o', edgecolors=color_m, facecolors='none', s=maker_size_mics, linewidths=LW)

size = 0.25
for rad_mult in [1,2,3]:
    ax.plot(rad_mult*size*np.cos(t), rad_mult*size*np.sin(t), c=color_n, linewidth=LW)
    ax.plot(rad_mult*size*np.cos(t)+room_length, rad_mult*size*np.sin(t), c=color_n, linewidth=LW)
    ax.plot(rad_mult*size*np.cos(t), rad_mult*size*np.sin(t)+room_width, c=color_n, linewidth=LW)
    ax.plot(rad_mult*size*np.cos(t)+room_length, rad_mult*size*np.sin(t)+room_width, c=color_n, linewidth=LW)

# Set aspect ratio to equal for equal visual increments
ax.set_aspect('equal')

# Make plot boundaries thicker
for spine in ax.spines.values():
    spine.set_linewidth(LW)

# Set larger font size for labels
# plt.xlabel(r'Room Length [m]', fontsize=label_fontsize)
# plt.ylabel(r'Room Width [m]', fontsize=label_fontsize)
plt.grid(False)
# Remove ticks and tick labels
ax.set_xticks([])
ax.set_yticks([])

# Create custom marker for legend with head shape
import matplotlib.path as mpath
import matplotlib.lines as mlines
circle = mpath.Path.unit_circle()
nose_verts = np.array([[0.9, -0.3], [1.5, 0.], [0.9, 0.3]])
nose_codes = [mpath.Path.MOVETO, mpath.Path.LINETO, mpath.Path.LINETO]
verts = np.concatenate([circle.vertices, nose_verts])
codes = np.concatenate([circle.codes, nose_codes])
head_marker = mpath.Path(verts, codes)

# Create a proxy artist for the legend
spkr_legend = mlines.Line2D([], [], color=color_s, marker=head_marker, 
                              linestyle='None', markersize=15, 
                              markeredgewidth=LW, markeredgecolor=color_s, 
                              markerfacecolor='none', label=r'Speakers')

handles, labels = ax.get_legend_handles_labels()
# Add custom speaker handle at the beginning
handles = [spkr_legend] + handles
labels = [h.get_label() for h in handles]


plt.legend(handles=handles, labels=labels, fontsize=legend_fontsize, loc='lower center')#, bbox_to_anchor=(0.5, 1))
plt.text(0.75, 5.5, r'Background Noise', fontsize=text_fontsize, c=color_n)
plt.show()

# Save the figure as a PDF for LaTeX
# fig.savefig('room_plot.pdf', bbox_inches='tight', pad_inches=0.01)
# fig.savefig('/data2/Henri/Journal3/framewiseSpeakerCounting/documents/ICASSP2026/figures/room_plot.pdf', bbox_inches='tight', pad_inches=0.01)
fig.savefig('room_plot.pdf', pad_inches=0.01, dpi=600)
# fig.savefig('/data2/Henri/Journal3/framewiseSpeakerCounting/documents/ICASSP2026/figures/room_plot.pdf', pad_inches=0.01, dpi=600)
# fig.savefig('/data2/Henri/Journal3/framewiseSpeakerCounting/documents/presentation/figures_new/room_plot.pdf', bbox_inches='tight', pad_inches=0.01, dpi=600)
```
