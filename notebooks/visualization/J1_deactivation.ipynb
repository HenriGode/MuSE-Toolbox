```python
# Now load the csv files and concatenate them into a single dataframe
import pandas as pd
dfs = {}
for key, resultsdir in resultsdirs.items():
    df_list = []
    for method_short, method_long in zip(method_names_short, method_names_long):
        csv_file = resultsdir / f"{method_long}_results.csv"
        if csv_file.exists():
            # note that the first colum is the scenario_id which should be the index of the df
            df = pd.read_csv(csv_file, index_col=0)
            df["method"] = method_short  # add a column for the method name
            # Append method name to index to ensure uniqueness across the combined dataframe
            df.index = df.index.astype(str) + "_" + method_short
            df_list.append(df)
        else:
            print(f"Warning: {csv_file} does not exist and will be skipped.")
    # Concatenate all dataframes
    df_all = pd.concat(df_list)
    print("Combined dataframe shape:", df_all.shape)
    # print an overview of all columns
    print("Columns in the combined dataframe:", df_all.columns)
    # For all columns with MHA in their name, they are Hermitian Angle metrics up to now in radians,
    # convert them to degrees for better interpretability
    mha_columns = [col for col in df_all.columns if "MHA" in col]
    for col in mha_columns:
        df_all[col] = df_all[col] * (180.0 / 3.141592653589793)  # radians to degrees
    dfs[key] = df_all
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
import matplotlib.patches as mpatches
import pandas as pd
import numpy as np
from pathlib import Path

# 1. defined dimensions from your LaTeX document
latex_two_column_width_pt = 516.0
latex_font_size_pt = 8.0

# 2. Convert to inches for matplotlib
inches_per_pt = 1.0 / 72
fig_width_in = latex_two_column_width_pt * inches_per_pt 
# Height: Approx half the height of the 2x2 grid, maintaining aspect ratio for a single row
fig_height_in = fig_width_in * ((5**0.5 - 1) / 2) * 0.55 

# 3. Update rcParams
mpl.use("pgf")
mpl.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
    'pgf.preamble': r'\usepackage{amsmath,graphicx,siunitx,bm}',
    'font.size': latex_font_size_pt,
    'axes.labelsize': latex_font_size_pt,      
    'axes.titlesize': latex_font_size_pt,      
    'legend.fontsize': latex_font_size_pt, 
    'xtick.labelsize': latex_font_size_pt, 
    'ytick.labelsize': latex_font_size_pt,
    'figure.figsize': [fig_width_in, fig_height_in],
})

def plot_sad_comparison_row(
    df_oracle,
    df_cosad,
    metric_col,
    ylabel_text,
    output_files,
):
    # Fixed method order matching your main experiment
    all_methods = ["GSS", "CWu", "BOP", "BOP-S", "BOP-W", "BOPO", "BOPO-S", "BOPO-W"]  # Removed "Oracle"
    all_methods = ["Baseline", "Proposed 1", "Proposed 2", "Proposed Combi"]
    
    # Create color mapping
    palette_colors = sns.color_palette("colorblind", n_colors=len(all_methods))
    method_color_map = dict(zip(all_methods, palette_colors))
    
    # Create figure (1 row, 2 columns)
    fig, axes = plt.subplots(1, 2, figsize=(fig_width_in, fig_height_in), dpi=600)
    
    letters = ["(a)", "(b)"]
    titles = [
        "Source extraction performance (oracle source activity)",
        "Source extraction performance (estimated source activity)"
        ]
    
    # Store min medians for shared scaling
    min_medians = []
    
    # --- 1. Left Plot: Oracle SAD ---
    ax_oracle = axes[0]
    
    # Calculate min median
    medians_oracle = df_oracle.groupby(["input_snr", "method"])[metric_col].median()
    if not medians_oracle.empty:
        min_medians.append(medians_oracle.min())
    else:
        min_medians.append(0)

    # Filter to ensure we only use defined methods
    df_oracle_plot = df_oracle[df_oracle["method"].isin(all_methods)]
    current_methods_oracle = [m for m in all_methods if m in df_oracle_plot["method"].unique()]

    sns.barplot(
        data=df_oracle_plot,
        x="input_snr",
        y=metric_col,
        hue="method",
        hue_order=current_methods_oracle,
        palette=method_color_map,
        errorbar="se",
        estimator=pd.Series.median,
        ax=ax_oracle,
        linewidth=0.15,
    )
    
    ax_oracle.grid(True, axis='y', linestyle='--', alpha=0.7, linewidth=0.5)
    ax_oracle.set_title(titles[0], pad=3)
    ax_oracle.set_ylabel(ylabel_text)
    ax_oracle.set_xlabel("Input SNR [dB]")
    ax_oracle.text(-0.08, 1.1, r"{" + letters[0] + "}", 
                   transform=ax_oracle.transAxes, va='top', ha='right', fontsize=latex_font_size_pt)
    
    # Remove default legend from Left Plot
    if ax_oracle.get_legend():
        ax_oracle.get_legend().remove()
        
    # --- 2. Right Plot: COSAD ---
    ax_cosad = axes[1]
    
    # Filter Data: Exclude GSS and Oracle explicitly
    df_cosad_plot = df_cosad[~df_cosad["method"].isin(["GSS", "Oracle"])]
    df_cosad_plot = df_cosad_plot[df_cosad_plot["method"].isin(all_methods)] # Safety check
    
    current_methods_cosad = [m for m in all_methods if m in df_cosad_plot["method"].unique()]
    
    # Calculate min median
    medians_cosad = df_cosad_plot.groupby(["input_snr", "method"])[metric_col].median()
    if not medians_cosad.empty:
        min_medians.append(medians_cosad.min())
    else:
        min_medians.append(0)

    sns.barplot(
        data=df_cosad_plot,
        x="input_snr",
        y=metric_col,
        hue="method",
        hue_order=current_methods_cosad,
        palette=method_color_map,
        errorbar="se",
        estimator=pd.Series.median,
        ax=ax_cosad,
        linewidth=0.15,
    )

    ax_cosad.grid(True, axis='y', linestyle='--', alpha=0.7, linewidth=0.5)
    ax_cosad.set_title(titles[1], pad=3)
    ax_cosad.set_ylabel(ylabel_text)
    ax_cosad.set_xlabel("Input SNR [dB]")
    ax_cosad.text(-0.08, 1.1, r"{" + letters[1] + "}", 
                   transform=ax_cosad.transAxes, va='top', ha='right', fontsize=latex_font_size_pt)
    
    # Remove auto legend before adding custom one
    if ax_cosad.get_legend():
        ax_cosad.get_legend().remove()

    # --- Custom Legend on Right Plot (COSAD) ---
    # Legend includes ALL methods, even if not in this specific plot
    legend_handles = [mpatches.Patch(color=method_color_map[m], label=m) for m in all_methods]
    # ax_cosad.legend(
    #     handles=legend_handles,
    #     loc='upper right', 
    #     ncol=3,                     
    #     frameon=True,
    #     columnspacing=0.5,
    #     handletextpad=0.2,
    #     labelspacing=0.2,
    #     fontsize=latex_font_size_pt,
    # )

    ax_cosad.legend(
        handles=legend_handles,
        loc='upper left', 
        ncol=2,                     
        frameon=True,
        columnspacing=0.5,
        handletextpad=0.2,
        labelspacing=0.2,
        fontsize=latex_font_size_pt,
    )

    # --- Shared Scaling ---
    # Min: min_median - 1, rounded down
    global_min = min(min_medians)
    y_min = np.floor(global_min - 1)
    
    # Max: Max of both existing limits
    y_max = max(ax_oracle.get_ylim()[1], ax_cosad.get_ylim()[1])
    
    ax_oracle.set_ylim(y_min, y_max)
    ax_cosad.set_ylim(y_min, y_max)
    
    # Align y-labels
    # ax_oracle.yaxis.set_label_coords(-0.13, 0.5)
    # ax_cosad.yaxis.set_label_coords(-0.13, 0.5)

    plt.tight_layout(pad=0.5, w_pad=1.5)
    
    if isinstance(output_files, list):
        for of in output_files:
            plt.savefig(of, bbox_inches='tight', pad_inches=0.02)
        plt.close()
    else:
        plt.savefig(output_files, bbox_inches='tight', pad_inches=0.02)
        plt.close()

def generate_deactivation_latex_table(df_oracle, df_cosad, output_files):
    methods = ["GSS", "CWu", "BOP", "BOP-S", "BOP-W", "BOPO", "BOPO-S", "BOPO-W"]
    
    method_labels = {
        "GSS": r"\gls{GSS}~\cite{horiguchi2021blockonlinegss}",
        "CWu": r"\gls{CWu}~\cite{markovich-golan_performance_2018}",
        "BOP": r"\gls{BOP}~\cite{cherkassky_successive_2020}",
        "BOP-S": r"\gls{BOP-S}*",
        "BOP-W": r"\gls{BOP-W}*",
        "BOPO": r"\gls{BOPO}*",
        "BOPO-S": r"\gls{BOPO-S}*",
        "BOPO-W": r"\gls{BOPO-W}*",
    }

    snr_levels = [0, 5, 10, 15]

    # Calculate medians
    medians_oracle = df_oracle.groupby(["input_snr", "method"])["DSINR"].median()
    medians_cosad = df_cosad.groupby(["input_snr", "method"])["DSINR"].median()

    # Helper for bolding
    def is_best(val, snr, medians_series, methods_subset):
            # Filter methods present in this SNR
            valid_methods = [m for m in methods_subset if (snr, m) in medians_series.index]
            if not valid_methods: return False
            vals = [medians_series.loc[(snr, m)] for m in valid_methods]
            best_val = max(vals) # best is highest DSINR
            return val >= best_val - 1e-6

    # Use 'table' instead of 'table*' as it is not too wide
    latex_str = r"""\begin{table}[htbp]
        \color{blue}
\centering
\begin{tabular}{p{2cm}cccc}
\toprule
\multirow{2}{*}{\textbf{Method}} & \multicolumn{4}{c}{\textbf{\gls{SNR}}} \\
 & 0 & 5 & 10 & 15 \\
\midrule
\multicolumn{5}{l}{\textbf{Oracle Source Activity}} \\
\midrule
"""
    # Section 1: Oracle
    methods_oracle = methods # All methods
    for method in methods_oracle:
        row_str = f"{method_labels[method]} & "
        for i, snr in enumerate(snr_levels):
            if (snr, method) in medians_oracle.index:
                val = medians_oracle.loc[(snr, method)]
                if is_best(val, snr, medians_oracle, methods_oracle):
                    val_str = r"\textbf{" + f"$\\SI{{{val:.1f}}}{{\\dB}}$" + "}"
                else:
                    val_str = f"$\\SI{{{val:.1f}}}{{\\dB}}$"
                row_str += val_str
            else:
                row_str += "-"
            if i < len(snr_levels) - 1:
                row_str += " & "
        latex_str += row_str + r" \\" + "\n"

    latex_str += r"""\midrule
\multicolumn{5}{l}{\textbf{Estimated Source Activity}} \\
\midrule
"""
    # Section 2: COSAD (No GSS)
    methods_cosad = [m for m in methods if m != "GSS"]
    for method in methods_cosad:
        row_str = f"{method_labels[method]} & "
        for i, snr in enumerate(snr_levels):
            if (snr, method) in medians_cosad.index:
                val = medians_cosad.loc[(snr, method)]
                if is_best(val, snr, medians_cosad, methods_cosad):
                    val_str = r"\textbf{" + f"$\\SI{{{val:.1f}}}{{\\dB}}$" + "}"
                else:
                    val_str = f"$\\SI{{{val:.1f}}}{{\\dB}}$"
                row_str += val_str
            else:
                row_str += "-"
            if i < len(snr_levels) - 1:
                row_str += " & "
        latex_str += row_str + r" \\" + "\n"
    
    latex_str += r"\bottomrule" + "\n"
    latex_str += r"\end{tabular}" + "\n"
    
    caption = r"Median \gls{SINR} improvement of all considered methods for different \glspl{SNR} using oracle and estimated source activity."
    latex_str += r"\vspace{0.1cm}" + "\n"
    latex_str += f"\\caption{{{caption}}}" + "\n"
    latex_str += r"\label{tab:deactivation_combined}" + "\n"
    latex_str += r"\end{table}" + "\n"

    # Write to files
    if isinstance(output_files, list):
        for out_file in output_files:
            tex_path = Path(out_file).with_suffix(".tex")
            with open(tex_path, "w") as f:
                f.write(latex_str)
            print(f"Latex table saved to {tex_path}")
    else:
        tex_path = Path(output_files).with_suffix(".tex")
        with open(tex_path, "w") as f:
            f.write(latex_str)
        print(f"Latex table saved to {tex_path}")


# Define Paths
output_pdf_deact = resultsdirs["OracleSAD"] / "plots" / "ScienceSlam_Deactivation_50ms_DSINR_Row.pdf"
paper_figure_path = Path("/data4/Henri/j3/framewiseSpeakerCounting/documents/Journal1/figures/Rebuttal")
paper_out_pdf_deact = paper_figure_path / "ScienceSlam_Deactivation_50ms_DSINR_Row.pdf"

# Ensure output dir exists
output_pdf_deact.parent.mkdir(parents=True, exist_ok=True)
paper_out_pdf_deact.parent.mkdir(parents=True, exist_ok=True)

# Run
print("Generating Deactivation Experiment Plot...")
plot_sad_comparison_row(
    df_oracle=dfs["OracleSAD"].reset_index(drop=True),
    df_cosad=dfs["COSAD"].reset_index(drop=True),
    metric_col="DSINR", 
    ylabel_text=r"SINR Improvement [dB]",
    output_files=[output_pdf_deact, paper_out_pdf_deact]
)
print(f"Saved to {output_pdf_deact}")

# Run Table Generation
print("Generating Deactivation LaTeX Table...")
generate_deactivation_latex_table(
    df_oracle=dfs["OracleSAD"].reset_index(drop=True),
    df_cosad=dfs["COSAD"].reset_index(drop=True),
    output_files=[output_pdf_deact, paper_out_pdf_deact]
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
import matplotlib.patches as mpatches
import pandas as pd
import numpy as np
from pathlib import Path

# 1. defined dimensions from your LaTeX document
latex_two_column_width_pt = 516.0
latex_font_size_pt = 8.0

# 2. Convert to inches for matplotlib
inches_per_pt = 1.0 / 72
fig_width_in = latex_two_column_width_pt * inches_per_pt 
# Height: Approx half the height of the 2x2 grid, maintaining aspect ratio for a single row
fig_height_in = fig_width_in * ((5**0.5 - 1) / 2) * 0.55 

# 3. Update rcParams
mpl.use("pgf")
mpl.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
    'pgf.preamble': r'\usepackage{amsmath,graphicx,siunitx,bm}',
    'font.size': latex_font_size_pt,
    'axes.labelsize': latex_font_size_pt,      
    'axes.titlesize': latex_font_size_pt,      
    'legend.fontsize': latex_font_size_pt, 
    'xtick.labelsize': latex_font_size_pt, 
    'ytick.labelsize': latex_font_size_pt,
    'figure.figsize': [fig_width_in, fig_height_in],
})

def plot_sad_comparison_row(
    df_oracle,
    df_cosad,
    metric_col,
    ylabel_text,
    output_files,
):
    # Fixed method order matching your main experiment
    all_methods = ["GSS", "CWu", "BOP", "BOP-S", "BOP-W", "BOPO", "BOPO-S", "BOPO-W"]  # Removed "Oracle"
    
    # Create color mapping
    palette_colors = sns.color_palette("colorblind", n_colors=len(all_methods))
    method_color_map = dict(zip(all_methods, palette_colors))
    
    # Create figure (1 row, 2 columns)
    fig, axes = plt.subplots(1, 2, figsize=(fig_width_in, fig_height_in), dpi=600)
    
    letters = ["(a)", "(b)"]
    titles = [
        "Source extraction performance (oracle source activity)",
        "Source extraction performance (estimated source activity)"
        ]
    
    # Store min medians for shared scaling
    min_medians = []
    
    # --- 1. Left Plot: Oracle SAD ---
    ax_oracle = axes[0]
    
    # Calculate min median
    medians_oracle = df_oracle.groupby(["input_snr", "method"])[metric_col].median()
    if not medians_oracle.empty:
        min_medians.append(medians_oracle.min())
    else:
        min_medians.append(0)

    # Filter to ensure we only use defined methods
    df_oracle_plot = df_oracle[df_oracle["method"].isin(all_methods)]
    current_methods_oracle = [m for m in all_methods if m in df_oracle_plot["method"].unique()]

    sns.barplot(
        data=df_oracle_plot,
        x="input_snr",
        y=metric_col,
        hue="method",
        hue_order=current_methods_oracle,
        palette=method_color_map,
        errorbar="se",
        estimator=pd.Series.median,
        ax=ax_oracle,
        linewidth=0.15,
    )
    
    ax_oracle.grid(True, axis='y', linestyle='--', alpha=0.7, linewidth=0.5)
    ax_oracle.set_title(titles[0], pad=3)
    ax_oracle.set_ylabel(ylabel_text)
    ax_oracle.set_xlabel("Input SNR [dB]")
    ax_oracle.text(-0.08, 1.1, r"{" + letters[0] + "}", 
                   transform=ax_oracle.transAxes, va='top', ha='right', fontsize=latex_font_size_pt)
    
    # Remove default legend from Left Plot
    if ax_oracle.get_legend():
        ax_oracle.get_legend().remove()
        
    # --- 2. Right Plot: COSAD ---
    ax_cosad = axes[1]
    
    # Filter Data: Exclude GSS and Oracle explicitly
    df_cosad_plot = df_cosad[~df_cosad["method"].isin(["GSS", "Oracle"])]
    df_cosad_plot = df_cosad_plot[df_cosad_plot["method"].isin(all_methods)] # Safety check
    
    current_methods_cosad = [m for m in all_methods if m in df_cosad_plot["method"].unique()]
    
    # Calculate min median
    medians_cosad = df_cosad_plot.groupby(["input_snr", "method"])[metric_col].median()
    if not medians_cosad.empty:
        min_medians.append(medians_cosad.min())
    else:
        min_medians.append(0)

    sns.barplot(
        data=df_cosad_plot,
        x="input_snr",
        y=metric_col,
        hue="method",
        hue_order=current_methods_cosad,
        palette=method_color_map,
        errorbar="se",
        estimator=pd.Series.median,
        ax=ax_cosad,
        linewidth=0.15,
    )

    ax_cosad.grid(True, axis='y', linestyle='--', alpha=0.7, linewidth=0.5)
    ax_cosad.set_title(titles[1], pad=3)
    ax_cosad.set_ylabel(ylabel_text)
    ax_cosad.set_xlabel("Input SNR [dB]")
    ax_cosad.text(-0.08, 1.1, r"{" + letters[1] + "}", 
                   transform=ax_cosad.transAxes, va='top', ha='right', fontsize=latex_font_size_pt)
    
    # Remove auto legend before adding custom one
    if ax_cosad.get_legend():
        ax_cosad.get_legend().remove()

    # --- Custom Legend on Right Plot (COSAD) ---
    # Legend includes ALL methods, even if not in this specific plot
    legend_handles = [mpatches.Patch(color=method_color_map[m], label=m) for m in all_methods]
    ax_cosad.legend(
        handles=legend_handles,
        loc='upper right', 
        ncol=3,                     
        frameon=True,
        columnspacing=0.5,
        handletextpad=0.2,
        labelspacing=0.2,
        fontsize=latex_font_size_pt,
    )

    # --- Shared Scaling ---
    # Min: min_median - 1, rounded down
    global_min = min(min_medians)
    y_min = np.floor(global_min - 1)
    
    # Max: Max of both existing limits
    y_max = max(ax_oracle.get_ylim()[1], ax_cosad.get_ylim()[1])
    
    if metric_col in ["STOIi", "STOIo", "DSTOI", "STOIi_A1", "STOIo_A1", "DSTOI_A1", "STOIi_A2", "STOIo_A2", "DSTOI_A2", "STOIi_A3", "STOIo_A3", "DSTOI_A3"]:
        y_min = 0.0
        y_max = 1.0
    
    ax_oracle.set_ylim(y_min, y_max)
    ax_cosad.set_ylim(y_min, y_max)
    
    # Align y-labels
    # ax_oracle.yaxis.set_label_coords(-0.13, 0.5)
    # ax_cosad.yaxis.set_label_coords(-0.13, 0.5)

    plt.tight_layout(pad=0.5, w_pad=1.5)
    
    if isinstance(output_files, list):
        for of in output_files:
            plt.savefig(of, bbox_inches='tight', pad_inches=0.02)
        plt.close()
    else:
        plt.savefig(output_files, bbox_inches='tight', pad_inches=0.02)
        plt.close()

def generate_deactivation_latex_table(df_oracle, df_cosad, metric_col, output_files):
    methods = ["GSS", "CWu", "BOP", "BOP-S", "BOP-W", "BOPO", "BOPO-S", "BOPO-W"]
    
    method_labels = {
        "GSS": r"\gls{GSS}~\cite{horiguchi2021blockonlinegss}",
        "CWu": r"\gls{CWu}~\cite{markovich-golan_performance_2018}",
        "BOP": r"\gls{BOP}~\cite{cherkassky_successive_2020}",
        "BOP-S": r"\gls{BOP-S}*",
        "BOP-W": r"\gls{BOP-W}*",
        "BOPO": r"\gls{BOPO}*",
        "BOPO-S": r"\gls{BOPO-S}*",
        "BOPO-W": r"\gls{BOPO-W}*",
    }

    snr_levels = [0, 5, 10, 15]

    # Calculate medians
    medians_oracle = df_oracle.groupby(["input_snr", "method"])[metric_col].median()
    medians_cosad = df_cosad.groupby(["input_snr", "method"])[metric_col].median()

    # Helper for bolding
    def is_best(val, snr, medians_series, methods_subset):
            # Filter methods present in this SNR
            valid_methods = [m for m in methods_subset if (snr, m) in medians_series.index]
            if not valid_methods: return False
            vals = [medians_series.loc[(snr, m)] for m in valid_methods]
            best_val = max(vals) # best is highest DSINR
            return val >= best_val - 1e-6

    # Use 'table' instead of 'table*' as it is not too wide
    latex_str = r"""\begin{table}[htbp]
\centering
\begin{tabular}{p{2cm}cccc}
\toprule
\multirow{2}{*}{\textbf{Method}} & \multicolumn{4}{c}{\textbf{\gls{SNR}}} \\
 & 0 & 5 & 10 & 15 \\
\midrule
\multicolumn{5}{l}{\textbf{Oracle Source Activity}} \\
\midrule
"""
    # Section 1: Oracle
    methods_oracle = methods # All methods
    for method in methods_oracle:
        row_str = f"{method_labels[method]} & "
        for i, snr in enumerate(snr_levels):
            if (snr, method) in medians_oracle.index:
                val = medians_oracle.loc[(snr, method)]
                if is_best(val, snr, medians_oracle, methods_oracle):
                    val_str = r"\textbf{" + f"$\\SI{{{val:.1f}}}{{\\dB}}$" + "}"
                else:
                    val_str = f"$\\SI{{{val:.1f}}}{{\\dB}}$"
                row_str += val_str
            else:
                row_str += "-"
            if i < len(snr_levels) - 1:
                row_str += " & "
        latex_str += row_str + r" \\" + "\n"

    latex_str += r"""\midrule
\multicolumn{5}{l}{\textbf{Estimated Source Activity}} \\
\midrule
"""
    # Section 2: COSAD (No GSS)
    methods_cosad = [m for m in methods if m != "GSS"]
    for method in methods_cosad:
        row_str = f"{method_labels[method]} & "
        for i, snr in enumerate(snr_levels):
            if (snr, method) in medians_cosad.index:
                val = medians_cosad.loc[(snr, method)]
                if is_best(val, snr, medians_cosad, methods_cosad):
                    val_str = r"\textbf{" + f"$\\SI{{{val:.1f}}}{{\\dB}}$" + "}"
                else:
                    val_str = f"$\\SI{{{val:.1f}}}{{\\dB}}$"
                row_str += val_str
            else:
                row_str += "-"
            if i < len(snr_levels) - 1:
                row_str += " & "
        latex_str += row_str + r" \\" + "\n"
    
    latex_str += r"\bottomrule" + "\n"
    latex_str += r"\end{tabular}" + "\n"
    
    caption = r"Median \gls{SINR} improvement of all considered methods for different \glspl{SNR} using oracle and estimated source activity."
    latex_str += r"\vspace{0.1cm}" + "\n"
    latex_str += f"\\caption{{{caption}}}" + "\n"
    latex_str += r"\label{tab:deactivation_combined}" + "\n"
    latex_str += r"\end{table}" + "\n"

    # Write to files
    if isinstance(output_files, list):
        for out_file in output_files:
            tex_path = Path(out_file).with_suffix(".tex")
            with open(tex_path, "w") as f:
                f.write(latex_str)
            print(f"Latex table saved to {tex_path}")
    else:
        tex_path = Path(output_files).with_suffix(".tex")
        with open(tex_path, "w") as f:
            f.write(latex_str)
        print(f"Latex table saved to {tex_path}")


metric = ("STOIo", r"STOI")

# Define Paths
output_pdf_deact = resultsdirs["OracleSAD"] / "plots" / f"Deactivation_50ms_{metric[0]}_Row.pdf"
paper_figure_path = Path("/data4/Henri/j3/framewiseSpeakerCounting/documents/Journal1/figures/Rebuttal")
paper_out_pdf_deact = paper_figure_path / f"Deactivation_50ms_{metric[0]}_Row.pdf"

# Ensure output dir exists
output_pdf_deact.parent.mkdir(parents=True, exist_ok=True)
paper_out_pdf_deact.parent.mkdir(parents=True, exist_ok=True)

# Run
print("Generating Deactivation Experiment Plot...")
plot_sad_comparison_row(
    df_oracle=dfs["OracleSAD"].reset_index(drop=True),
    df_cosad=dfs["COSAD"].reset_index(drop=True),
    metric_col=metric[0],
    ylabel_text=metric[1],
    output_files=[output_pdf_deact, paper_out_pdf_deact]
)
print(f"Saved to {output_pdf_deact}")

# Run Table Generation
print("Generating Deactivation LaTeX Table...")
generate_deactivation_latex_table(
    df_oracle=dfs["OracleSAD"].reset_index(drop=True),
    df_cosad=dfs["COSAD"].reset_index(drop=True),
    metric_col=metric[0],
    output_files=[output_pdf_deact, paper_out_pdf_deact]
)
```
